What Is Machine Learning?

Bu bob Machine Learning (Mashinali o‘rganish) haqida. Machine Learning — bu kompyuterga ma'lumotlardan o‘rganish va ular asosida kelajakdagi natijalarni bashorat qilish imkonini beradigan usullar to‘plamidir. Unda matematik modellar va algoritmlar yordamida ma'lumotlar tahlil qilinadi. Bob davomida Scikit-Learn kutubxonasi yordamida Machine Learning usullari amaliy misollar bilan o‘rganiladi.

| Atama                                       | O'zbekcha ma'nosi          | Qisqacha izoh                                                                             |
| ------------------------------------------- | -------------------------- | ----------------------------------------------------------------------------------------- |
| **Machine Learning (ML)**                   | Mashinali o‘rganish        | Kompyuterning ma'lumotlardan o‘rganishi va bashorat qilishi.                              |
| **Data Science**                            | Ma'lumotlar ilmi           | Ma'lumotlarni yig‘ish, tahlil qilish va ulardan foydali xulosa chiqarish sohasi.          |
| **Artificial Intelligence (AI)**            | Sun'iy intellekt           | Inson aqlini taqlid qiluvchi tizimlar va dasturlar.                                       |
| **Model**                                   | Model                      | Ma'lumotlar asosida yaratilgan matematik ifoda yoki algoritm.                             |
| **Mathematical Model**                      | Matematik model            | Muammoni matematik formulalar orqali ifodalash usuli.                                     |
| **Parameter**                               | Parametr                   | Modelning o‘rganish jarayonida o‘zgaradigan qiymati.                                      |
| **Tunable Parameters**                      | Sozlanadigan parametrlar   | Ma'lumotlarga qarab avtomatik moslashtiriladigan parametrlar.                             |
| **Learning**                                | O‘rganish                  | Modelning ma'lumotlardan qonuniyatlarni o‘zlashtirish jarayoni.                           |
| **Prediction**                              | Bashorat                   | O‘rganilgan model yordamida yangi natijalarni oldindan aytish.                            |
| **Inference**                               | Xulosa chiqarish           | Ma'lumotlardan mantiqiy yoki statistik natija olish.                                      |
| **Data Exploration**                        | Ma'lumotlarni o‘rganish    | Ma'lumotlarni tahlil qilib, ularning xususiyatlarini aniqlash.                            |
| **Bias**                                    | Og‘ish                     | Modelning noto‘g‘ri yoki soddalashtirilgan taxmin qilishi natijasidagi xatolik.           |

| **Training Data**                           | O‘quv ma'lumotlari         | Modelni o‘qitish uchun ishlatiladigan ma'lumotlar.                                        |
| **Observed Data**                           | Kuzatilgan ma'lumotlar     | Model o‘rganadigan mavjud ma'lumotlar.                                                    |
| **New Data**                                | Yangi ma'lumotlar          | Model bashorat qiladigan yoki tahlil qiladigan yangi ma'lumotlar.                         |


Categories of Machine Learning


| Atama                        | O‘zbekcha ma'nosi              | Sodda tushuntirish                                                      |
| ---------------------------- | ------------------------------ | ----------------------------------------------------------------------- |
| **Machine Learning**         | Mashinali o‘rganish            | Kompyuterning ma'lumotlardan o‘rganishi.                                |
| **Supervised Learning**      | Nazorat ostida o‘qitish        | Ma'lumot ham, to‘g‘ri javob (label) ham beriladi.                       |
| **Unsupervised Learning**    | Nazoratsiz o‘qitish            | Faqat ma'lumot beriladi, model o‘zi qonuniyat topadi.                   |
| **Semi-Supervised Learning** | Qisman nazorat ostida o‘qitish | Ba'zi ma'lumotlarda label bor, ba'zilarida yo‘q.                        |
| **Feature**                  | Xususiyat                      | Model o‘rganadigan ma'lumot ustuni (masalan, yosh, vazn).               |
| **Label**                    | Yorliq, javob                  | Model topishi kerak bo‘lgan natija.                                     |
| **Classification**           | Klassifikatsiya                | Natija kategoriya bo‘ladi (Ha/Yo‘q, Mushuk/It).                         |
| **Regression**               | Regressiya                     | Natija son bo‘ladi (narx, harorat, maosh).                              |
| **Clustering**               | Klasterlash                    | O‘xshash ma'lumotlarni guruhlarga ajratish.                             |
| **Dimensionality Reduction** | O‘lchamni kamaytirish          | Keraksiz xususiyatlarni qisqartirish orqali ma'lumotni soddalashtirish. |


| **Machine Learning turi**                            | **Asosiy savol**                                    | **Qisqacha tushuntirish**                                                        | **Misollar**                                                                          |
| ---------------------------------------------------- | --------------------------------------------------- | -------------------------------------------------------------------------------- | ------------------------------------------------------------------------------------- |
| **Classification (Klassifikatsiya)**                 | **Qaysi guruh?**                                    | Ma'lumotni oldindan belgilangan kategoriyalardan biriga ajratadi.                | Spam yoki Spam emas, Mushuk yoki It, Kasal yoki Sog‘lom                               |
| **Regression (Regressiya)**                          | **Qancha?**                                         | Son qiymatini bashorat qiladi.                                                   | Uy narxi, Maosh, Harorat, Mashina narxi                                               |
| **Clustering (Klasterlash)**                         | **Qaysi guruhga o‘xshaydi?**                        | O‘xshash ma'lumotlarni avtomatik ravishda guruhlarga ajratadi.                   | Mijozlarni xarid odatiga qarab guruhlash, Talabalarni natijalariga qarab ajratish     |
| **Dimensionality Reduction (O‘lchamni kamaytirish)** | **Qanday qilib ma'lumotni soddalashtirish mumkin?** | Feature (xususiyat)lar sonini kamaytirib, eng muhim ma'lumotlarni saqlab qoladi. | 1000 ta feature → 2 yoki 3 ta featurega qisqartirish, Vizualizatsiyani osonlashtirish |


Supervised Learning formulasi


Feature + Label
-> 
Model o'rganadi
->
Yangi ma'lumot
->
 Prediction


Unsupervised formulasi

Feature
->
Model o'zi tahlil qiladi
->
O'xshashlarini topadi

Bir jumlada eslab qoling:

Supervised Learning → "Ustoz bilan o‘rganish." (Ma'lumot + javob beriladi.)

Unsupervised Learning → "Mustaqil o‘rganish." (Faqat ma'lumot beriladi, model o‘zi qonuniyatlarni topadi.)
