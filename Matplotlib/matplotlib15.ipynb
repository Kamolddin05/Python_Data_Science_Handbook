```python
data = np.random.multivariate_normal([0, 0], [[5, 2], [2, 2]], size=2000)
data = pd.DataFrame(data, columns=['x', 'y'])

for col in 'xy':
    plt.hist(data[col], alpha=0.5)
```

```python
for col in 'xy':
    sns.kdeplot(data[col], shade=True)

```

```python
sns.distplot(data['x'])
sns.distplot(data['y']);
```

```python
sns.kdeplot(data)
# Ikki o‘zgaruvchi orasidagi zichlikni ko‘rsatadi.
```

```python
with sns.axes_style('white'):
    sns.jointplot(data=data, x="x", y="y", kind="kde")

# Bir vaqtning o‘zida umumiy va alohida taqsimotni ko‘rsatadi.
```

```python
with sns.axes_style('white'):
    sns.jointplot(x = "x", y = "y", data = data, kind='hex')

# Ma'lumotlarni olti burchakli (hexagon) ko‘rinishda tasvirlaydi.
```

```python
# Ma'lumotlarni guruhlarga ajratib alohida histogramlar hosil qiladi.

tips = sns.load_dataset('tips')
tips.head()

tips['tip_pct'] = 100 * tips['tip'] / tips['total_bill']

grid = sns.FacetGrid(tips, row="sex", col="time", margin_titles=True)
grid.map(plt.hist, "tip_pct", bins=np.linspace(0, 40, 15))
```

```python
# Kategoriyalar bo‘yicha ma'lumotlarni taqqoslaydi.
import seaborn as sns

with sns.axes_style(style='ticks'):
    g = sns.catplot(data=tips, x="day", y="total_bill", hue="sex", kind="box")
    g.set_axis_labels("Day", "Total Bill")
```

```python
with sns.axes_style('white'):
    sns.jointplot(data=tips, x="total_bill", y="tip", kind="hex")
#Ikki ustun orasidagi bog‘lanishni ko‘rsatadi.
```

```python
planets = sns.load_dataset('planets')
planets.head()
```

```python
import seaborn as sns

with sns.axes_style('ticks'):
    g = sns.catplot(data=planets, x="year", aspect=2,
                    kind="count", color='steelblue')
    g.set_xticklabels(step=5)
```

```python
import seaborn as sns

with sns.axes_style('white'):
    g = sns.catplot(
        data=planets,
        x="year",
        aspect=4.0,
        kind="count",
        hue="method",
        order=range(2001, 2015),
        color='steelblue'
    )
    g.set_ylabels('Number of Planets Discovered')
```

| **Funksiya**       | **Sintaksisi (qanday yoziladi)**                    | **Vazifasi**                                                  |
| ------------------ | --------------------------------------------------- | ------------------------------------------------------------- |
| `plt.hist()`       | `plt.hist(data)`                                    | Histogram (taqsimot grafigi) chizadi.                         |
| `sns.kdeplot()`    | `sns.kdeplot(data)`                                 | Silliq zichlik grafigini (KDE) chizadi.                       |
| `sns.distplot()`   | `sns.distplot(data)`                                | Histogram va KDE ni birgalikda chiqaradi.                     |
| `sns.jointplot()`  | `sns.jointplot(x="x", y="y", data=df, kind="kde")`  | Ikki o‘zgaruvchi orasidagi bog‘lanishni ko‘rsatadi.           |
| `sns.pairplot()`   | `sns.pairplot(df, hue="column")`                    | DataFrame ustunlari orasidagi barcha bog‘lanishlarni chizadi. |
| `sns.FacetGrid()`  | `grid = sns.FacetGrid(df, row="col1", col="col2")`  | Ma'lumotlarni guruhlarga ajratib alohida grafiklar yaratadi.  |
| `sns.factorplot()` | `sns.factorplot(x="x", y="y", data=df, kind="box")` | Kategoriyalarni taqqoslaydi (box, count va boshqalar).        |

```python
import pandas as pd
from datetime import timedelta

def convert_time(s):
    h, m, s = map(int, s.split(':'))
    return timedelta(hours=h, minutes=m, seconds=s)

data = pd.read_csv(
    'marathon-data.csv',
    converters={'split': convert_time, 'final': convert_time}
)

print(data.head())
```

```python
data['split_sec'] = data['split'].astype(int) / 1E9
data['final_sec'] = data['final'].astype(int) / 1E9

data.head()
```

```python
import seaborn as sns
import numpy as np

with sns.axes_style('white'):
    g = sns.jointplot(data=data, x="split_sec", y="final_sec", kind="hex")

    g.ax_joint.plot(
        np.linspace(4000, 16000),
        np.linspace(8000, 32000),
        ':k'
    )
```

```python
data['split_sec'] = data['split'].astype(int) / 1E9
data['final_sec'] = data['final'].astype(int) / 1E9

data.head()
```

```python
import seaborn as sns
import numpy as np

with sns.axes_style('white'):
    g = sns.jointplot(data=data, x="split_sec", y="final_sec", kind="hex")

    g.ax_joint.plot(
        np.linspace(4000, 16000),
        np.linspace(8000, 32000),
        ':k'
    )
```

```python
data['split_frac'] = 1 - 2 * data['split_sec'] / data['final_sec']

data.head()
```

```python
sns.distplot(data['split_frac'], kde=False)

plt.axvline(0, color="k", linestyle="--")
```

```python
sum(data.split_frac < 0)
```

```python
g = sns.PairGrid(data,
                 vars=['age', 'split_sec', 'final_sec', 'split_frac'],
                 hue='gender',
                 palette='RdBu_r')

g.map(plt.scatter, alpha=0.8)
g.add_legend();
```

```python
sns.kdeplot(data.split_frac[data.gender=='M'],
            label='men',
            shade=True)

sns.kdeplot(data.split_frac[data.gender=='W'],
            label='women',
            shade=True)

plt.xlabel('split_frac');
```

```python
import seaborn as sns

sns.violinplot(
    x="gender",
    y="split_frac",
    data=data,
    palette=["lightblue", "lightpink"]
)
```

```python
data['age_dec'] = data.age.map(lambda age: 10 * (age // 10))

data.head()
```

```python
import seaborn as sns

with sns.axes_style('white'):
    g = sns.violinplot(
        data=data,
        x="age_dec",
        y="split_frac",
        hue="gender",
        split=True,
        inner="quartile",
        palette=["lightblue", "lightpink"]
    )
```

```python
(data.age > 80).sum()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

g = sns.lmplot(
    data=data,
    x="final_sec",
    y="split_frac",
    col="gender",
    markers=".",
    scatter_kws=dict(color="c")
)

g.map(plt.axhline, y=0.1, color="k", ls=":")
```

| **Grafik (Funksiya)** | **Sintaksisi (qanday yoziladi)**                                               | **Vazifasi**                                                           |

| `g.ax_joint.plot()`   | `g.ax_joint.plot(x, y, ':k')`                                                  | JointPlot ustiga qo‘shimcha chiziq (ideal chiziq) chizadi.             |

| `sns.PairGrid()`      | `sns.PairGrid(data, vars=[...], hue='gender')`                                 | Bir nechta ustunlar orasidagi barcha bog‘lanishlarni ko‘rsatadi.       |
| `g.map(plt.scatter)`  | `g.map(plt.scatter, alpha=0.8)`                                                | PairGrid ichidagi nuqtali (Scatter) grafiklarni chizadi.               |
| `sns.kdeplot()`       | `sns.kdeplot(data.split_frac[data.gender=='M'], shade=True)`                   | Erkak va ayollar taqsimotining silliq zichlik (KDE) grafigini chizadi. |
| `sns.violinplot()`    | `sns.violinplot("gender", "split_frac", data=data)`                            | Taqsimotni skripka (Violin) ko‘rinishida taqqoslaydi.                  |
| `sns.violinplot()`    | `sns.violinplot("age_dec", "split_frac", hue="gender", data=data, split=True)` | Yosh va jins bo‘yicha taqsimotni Violin grafikda ko‘rsatadi.           |

| `g.map(plt.axhline)`  | `g.map(plt.axhline, y=0.1, color='k', ls=':')`                                 | Regression grafiklariga gorizontal chiziq qo‘shadi.                    |
