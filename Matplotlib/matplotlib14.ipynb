```python
import matplotlib.pyplot as plt
from mpl_toolkits.basemap import Basemap

plt.figure(figsize=(12, 8), dpi=300)

m = Basemap(
    projection="merc",
    llcrnrlat=36,
    urcrnrlat=46,
    llcrnrlon=55,
    urcrnrlon=75,
    resolution="i"
)

# Fon
m.shadedrelief(scale=0.3)

# Davlat chegaralari
m.drawcountries(color="black", linewidth=1.2)

# Sohil chiziqlari
m.drawcoastlines(color="gray", linewidth=0.5)

# Parallel va meridianlar
m.drawparallels(range(36, 47, 2), labels=[1,0,0,0], fontsize=9)
m.drawmeridians(range(55, 76, 2), labels=[0,0,0,1], fontsize=9)

# Toshkent koordinatasi
x, y = m(69.2401, 41.2995)

# Nuqta
plt.plot(
    x, y,
    marker="o",
    color="red",
    markersize=8,
    markeredgecolor="white",
    markeredgewidth=1.5
)

# Yozuv
plt.text(
    x + 20000,
    y + 20000,
    "Tashkent",
    fontsize=11,
    color="white",
    weight="bold"
)

plt.title("Uzbekistan Map", fontsize=18, weight="bold")

plt.show()
```

```python
# Map Projections
```

```python
from itertools import chain

def draw_map(m, scale=0.2):
    m.shadedrelief(scale=scale)

    lats = m.drawparallels(np.linspace(-90, 90, 13))
    lons = m.drawmeridians(np.linspace(-180, 180, 13))

    lat_lines = chain(*(tup[1][0] for tup in lats.items()))
    lon_lines = chain(*(tup[1][0] for tup in lons.items()))

    all_lines = chain(lat_lines, lon_lines)

    for line in all_lines:
        line.set(linestyle='-', alpha=0.3, color='w')
```

```python
fig = plt.figure(figsize=(8,6))

m = Basemap(
    projection='cyl',
    resolution=None,
    llcrnrlat=-90,
    urcrnrlat=90,
    llcrnrlon=-180,
    urcrnrlon=180
)

draw_map(m)
```

```python
fig = plt.figure(figsize=(8,6))

m = Basemap(
    projection='moll',
    resolution=None,
    lat_0=0,
    lon_0=0
)

draw_map(m)
```

```python
fig = plt.figure(figsize=(8,8))

m = Basemap(
    projection='ortho',
    resolution=None,
    lat_0=50,
    lon_0=0
)

draw_map(m)
```

```python
fig = plt.figure(figsize=(8,8))

m = Basemap(
    projection='lcc',
    resolution=None,
    lon_0=0,
    lat_0=50,
    lat_1=45,
    lat_2=55,
    width=1.6E7,
    height=1.2E7
)

draw_map(m)
```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.basemap import Basemap

# 1 qator, 2 ustun
fig, ax = plt.subplots(1, 2, figsize=(12, 8), dpi=150)

# Low va Intermediate resolution
for i, res in enumerate(["l", "i"]):

    m = Basemap(
        projection="gnom",
        lat_0=57.3,
        lon_0=-6.2,
        width=90000,
        height=120000,
        resolution=res,
        ax=ax[i]
    )

    # Qit'alarni bo'yash
    m.fillcontinents(
        color="#FFDDCC",
        lake_color="#DDEEFF"
    )

    # Okean
    m.drawmapboundary(fill_color="#DDEEFF")

    # Qirg'oq chiziqlari
    m.drawcoastlines(
        color="black",
        linewidth=0.8
    )

    # Davlat chegaralari
    m.drawcountries(
        color="gray",
        linewidth=0.5
    )

    # Parallel chiziqlar
    m.drawparallels(
        range(-90, 91, 10),
        labels=[1, 0, 0, 0],
        fontsize=8
    )

    # Meridian chiziqlar
    m.drawmeridians(
        range(-180, 181, 10),
        labels=[0, 0, 0, 1],
        fontsize=8
    )

    ax[i].set_title(f"Resolution = '{res}'")

plt.suptitle("Basemap Resolution Comparison", fontsize=16)
plt.tight_layout()

plt.show()
```

```python
# Example: Surface Temperature Data
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.basemap import Basemap
from netCDF4 import Dataset, date2index
from datetime import datetime

# NetCDF faylini ochish
data = Dataset("gistemp250.nc")

# 2014-yil 15-yanvar indeksini olish
timeindex = date2index(
    datetime(2014, 1, 15),
    data.variables["time"]
)

# Latitude va Longitude
lat = data.variables["lat"][:]
lon = data.variables["lon"][:]

# 2D koordinatalar
lon, lat = np.meshgrid(lon, lat)

# Harorat anomaliyasi
temp_anomaly = data.variables["tempanomaly"][timeindex]

# Figure
plt.figure(figsize=(10, 8), dpi=150)

# Xarita
m = Basemap(
    projection="lcc",
    resolution="c",
    width=8E6,
    height=8E6,
    lat_0=45,
    lon_0=-100
)

# Fon
m.shadedrelief(scale=0.5)

# Haroratni chizish
mesh = m.pcolormesh(
    lon,
    lat,
    temp_anomaly,
    latlon=True,
    cmap="RdBu_r"
)

# Rang oralig'i
plt.clim(-8, 8)

# Qirg'oq chiziqlari
m.drawcoastlines(color="black", linewidth=0.5)

# Davlat chegaralari
m.drawcountries(color="gray")

# Parallel chiziqlar
m.drawparallels(
    range(-90, 91, 10),
    labels=[1, 0, 0, 0]
)

# Meridian chiziqlari
m.drawmeridians(
    range(-180, 181, 20),
    labels=[0, 0, 0, 1]
)

# Sarlavha
plt.title("January 2014 Temperature Anomaly", fontsize=16)

# Colorbar
plt.colorbar(
    mesh,
    label="Temperature Anomaly (°C)"
)

plt.show()
```
