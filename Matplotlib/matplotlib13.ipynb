```python
ax = plt.axes(projection = "3d")
ax.plot_surface(X, Y, Z, rstride = 1, cstride = 1, cmap = "viridis", edgecolor = "none" )
ax.set_title("surface")
plt.show()
```

```python
theta = np.linspace(0, 2*np.pi, 30)
w = np.linspace(-0.25, 0.25, 8)
w, theta = np.meshgrid(w, theta)
```

```python
phi = 0.5 * theta
```

```python
r = 1 + w * np.cos(phi)

x = np.ravel(r * np.cos(theta))
y = np.ravel(r * np.sin(theta))
z = np.ravel(w * np.sin(phi))
```

```python
from matplotlib.tri import Triangulation

tri = Triangulation(np.ravel(w), np.ravel(theta))

ax = plt.axes(projection='3d')
ax.plot_trisurf(
                x, y, z,
                triangles=tri.triangles,
                cmap='viridis',
                linewidths=0.2
                )

ax.set_xlim(-1, 1)
ax.set_ylim(-1, 1)
ax.set_zlim(-1, 1)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.tri import Triangulation

# Funksiya
def f(x, y):
    return np.sin(np.sqrt(x**2 + y**2))

# Tasodifiy nuqtalar
theta = 2 * np.pi * np.random.random(1000)
r = 6 * np.random.random(1000)

x = r * np.sin(theta)
y = r * np.cos(theta)
z = f(x, y)

# Triangulyatsiya
tri = Triangulation(x, y)

# Grafik
fig = plt.figure(figsize=(8,6))
ax = plt.axes(projection="3d")

ax.plot_trisurf(
    x,
    y,
    z,
    triangles=tri.triangles,
    cmap="viridis",
    edgecolor="none"
)

ax.set_title("3D Triangulated Surface")
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

plt.show()
```

| Funksiya                    | Vazifasi                                   |
| --------------------------- | ------------------------------------------ |
| `plt.axes(projection="3d")` | 3D koordinata yaratadi                     |
| `plot3D()`                  | 3D chiziq chizadi                          |
| `scatter3D()`               | 3D nuqtalar chizadi                        |
| `contour3D()`               | 3D kontur grafik chizadi                   |
| `plot_wireframe()`          | Simli karkas chizadi                       |
| `plot_surface()`            | To'ldirilgan 3D sirt chizadi               |
| `Triangulation()`           | Nuqtalarni uchburchaklarga ajratadi        |
| `plot_trisurf()`            | Uchburchaklar asosida 3D sirt chizadi      |
| `view_init()`               | Grafikning ko'rish burchagini o'zgartiradi |


| Funksiya           | Kalit so'z                     |
| ------------------ | ------------------------------ |
| `plot3D()`         | **Chiziq**                     |
| `scatter3D()`      | **Nuqta**                      |
| `contour3D()`      | **Kontur**                     |
| `plot_wireframe()` | **Karkas**                     |
| `plot_surface()`   | **To'ldirilgan sirt**          |
| `plot_trisurf()`   | **Tasodifiy nuqtalardan sirt** |


```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.tri import Triangulation

# ---------- Funksiya ----------
def f(x, y):
    return np.sin(np.sqrt(x**2 + y**2))

fig = plt.figure(figsize=(18, 10))

# ==========================================================
# 1. plot3D() - Chiziq
# ==========================================================
ax1 = fig.add_subplot(231, projection="3d")

z = np.linspace(0, 10, 300)
x = np.sin(z)
y = np.cos(z)

ax1.plot3D(x, y, z, color="blue")
ax1.set_title("plot3D() - Chiziq")


# ==========================================================
# 2. scatter3D() - Nuqtalar
# ==========================================================
ax2 = fig.add_subplot(232, projection="3d")

x = np.random.uniform(-2, 2, 300)
y = np.random.uniform(-2, 2, 300)
z = np.random.uniform(-2, 2, 300)

ax2.scatter3D(x, y, z, c=z, cmap="viridis")
ax2.set_title("scatter3D() - Nuqta")


# ==========================================================
# 3. contour3D() - Kontur
# ==========================================================
ax3 = fig.add_subplot(233, projection="3d")

x = np.linspace(-6, 6, 40)
y = np.linspace(-6, 6, 40)

X, Y = np.meshgrid(x, y)
Z = f(X, Y)

ax3.contour3D(X, Y, Z, 50, cmap="rainbow")
ax3.set_title("contour3D() - Kontur")


# ==========================================================
# 4. plot_wireframe() - Karkas
# ==========================================================
ax4 = fig.add_subplot(234, projection="3d")

x = np.linspace(-3, 3, 30)
y = np.linspace(-3, 3, 30)

X, Y = np.meshgrid(x, y)
Z = np.sin(X) * np.cos(Y)

ax4.plot_wireframe(X, Y, Z, color="black")
ax4.set_title("plot_wireframe() - Karkas")


# ==========================================================
# 5. plot_surface() - To'ldirilgan sirt
# ==========================================================
ax5 = fig.add_subplot(235, projection="3d")

ax5.plot_surface(
    X,
    Y,
    Z,
    cmap="viridis",
    edgecolor="none"
)

ax5.set_title("plot_surface() - To'ldirilgan sirt")


# ==========================================================
# 6. plot_trisurf() - Tasodifiy nuqtalardan sirt
# ==========================================================
ax6 = fig.add_subplot(236, projection="3d")

theta = 2 * np.pi * np.random.random(500)
r = 6 * np.random.random(500)

x = r * np.sin(theta)
y = r * np.cos(theta)
z = f(x, y)

tri = Triangulation(x, y)

ax6.plot_trisurf(
    x,
    y,
    z,
    triangles=tri.triangles,
    cmap="viridis",
    edgecolor="none"
)

ax6.set_title("plot_trisurf() - Triangulyatsiya")

plt.tight_layout()
plt.show()
```
